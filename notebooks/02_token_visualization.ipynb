# Token Visualization: PCA of Visual Tokens

Visualize the 576 CLIP visual tokens as RGB spatial maps using PCA.

**Goals:**
1. Extract 576 visual tokens from CLIP ViT-L/14 [576, 1024]
2. Apply PCA to reduce 1024-dim to 3-dim (RGB)
3. Reshape to 24×24 spatial grid
4. Visualize as RGB image to understand feature distribution

**Insight:** Shows which spatial regions have distinct visual features.
